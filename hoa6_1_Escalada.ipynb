```python
#Exercise 1 & 2
#Run the given code below and perform the given tasks without any python modules
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
```

**Exercise 1. Using the data generated above, calculate the following statistics without importing anything from the statistics module in the standard library**

```python

#1. Mean
mean = sum(salaries) / len(salaries)
print("Mean:", {mean})
#2. Median
sort = sorted(salaries)
n = len(sort)
median = (sort[n//2-1] + sort[n//2]) / 2
print("Median:", {median})
#3. Mode
map_values = {}
for salary in salaries:
  if salary not in map_values:
    map_values[salary] = 1
    continue
  map_values[salary] += map_values[salary] + 1

mode = max(map_values, key=map_values.get)
print("Mode:", {mode}, "Repeated", max(map_values.values()), "times")
```

```python
#4. Sample Variance
n = len(salaries)
sumresult = sum((x - mean)**2 for x in salaries)
variance = sumresult / (n - 1)
print("Sample Variance:", {variance})
#5. Sample Standard Deviation
standard_deviation = variance ** 0.5
print("Sample Standard Deviation:", {standard_deviation})
```

**Exercise 2.
Using the same data, calculate the following statistics using the functions in the statistics module where appropriate**

```python

import statistics

#1. Range
range = max(salaries) - min(salaries)
print(f"Range: {max(salaries)} - {min(salaries)} = {range}")
```

```python

#2. Interquartile Range
q1, q2, q3 = statistics.quantiles(salaries, n=4)
iqr = q3 - q1

#3. Coefficient of Variation
coef_variation = (statistics.stdev(salaries) / statistics.mean(salaries)) * 100

#4. Quartile Coefficient of Dispersion
qcd = (q3 - q1) / (q3 + q1)

print("Interquartile Range:", {iqr})
print("Coefficient of Variation:", {coef_variation})
print("Quartile Coefficient of Dispersion:", {qcd})
```
