```python
import pandas as pd
import numpy as np
```

```python
#Exercise 4
meteor_csv = pd.read_csv('/content/Meteorite_Landings.csv')

#Hint: Be sure to convert the year column to a number as we did in the previous exercise.
meteor_csv['year'] = meteor_csv['year'].str.slice(6, 10)
meteor_csv['year'] = pd.to_numeric(meteor_csv['year']).astype('Int64')
meteor_csv.head()
```

```python
#create a pivot table that shows both the number of meteorites and the 95th percentile of meteorite mass for those that were found
#versus observed falling per year from 2005 through 2009 (inclusive)
meteor_csv['mass (g)'] = pd.to_numeric(meteor_csv['mass (g)'])

observed_year = meteor_csv[
  meteor_csv['year'].between(2005,2009) & meteor_csv['fall'].isin(['Found', 'Fell'])
]

pivot_table = pd.pivot_table(observed_year, index='year', columns='fall', values='mass (g)', aggfunc=['count', lambda s: s.quantile(0.95)])

pivot_table.columns = [
    f"{'meteorite_count' if stat == 'count' else 'mass_95th_pct_g'}_{fall_type.lower()}"
    for stat, fall_type in pivot_table.columns
]

pivot_table
```

```python
#Meteorite and meteor masses that were found
found_data = meteor_csv[meteor_csv['fall']=='Found']
found_data = found_data.sort_index()
found_data
```

```python
#Meteorite and meteor masses that fell
fell_data = meteor_csv[meteor_csv['fall']=='Fell']
fell_data = fell_data.sort_index()
fell_data
```

```python
#Using the meteorite data from the Meteorite_Landings.csv file, compare summary statistics of the mass column for the meteorites that were found versus observed falling.

#Finding the mean, median, mode, variance, and deviation
found_data_mean = np.mean(found_data['mass (g)'])
found_data_median = np.median(found_data['mass (g)'])
found_data_mode = np.array(found_data['mass (g)'].mode())
found_data_var = np.var(found_data['mass (g)'])
found_data_std = np.std(found_data['mass (g)'])


fell_data_mean = np.mean(fell_data['mass (g)'])
fell_data_median = np.median(fell_data['mass (g)'])
fell_data_mode = np.array(fell_data['mass (g)'].mode())
fell_data_var = np.var(fell_data['mass (g)'])
fell_data_std = np.std(fell_data['mass (g)'])

compared_stats = pd.DataFrame({
    'fall_type': ['Found', 'Fell'],
    'mean_mass': [found_data_mean, fell_data_mean],
    'median_mass': [found_data_median, fell_data_median],
    'mode_mass': [found_data_mode, fell_data_mode],
    'variance_mass': [found_data_var, fell_data_var],
    'std_mass': [found_data_std, fell_data_std]
})

compared_stats
```

```python
#Exercise Part 5

#Taxi data: hourly totals and top 5 hours by tip amount
taxi = pd.read_csv('/content/2019_Yellow_Taxi_Trip_Data.csv')

dropoff_col_candidates = ['tpep_dropoff_datetime', 'dropoff_datetime', 'lpep_dropoff_datetime']
dropoff_col = next((c for c in dropoff_col_candidates if c in taxi.columns), None)
if dropoff_col is None:
    raise KeyError('No dropoff datetime column found in taxi dataset.')
taxi[dropoff_col] = pd.to_datetime(taxi[dropoff_col], errors='coerce')


sum_cols = ['trip_distance', 'fare_amount', 'tolls_amount', 'tip_amount']
for col in sum_cols:
    taxi[col] = pd.to_numeric(taxi[col], errors='coerce')


hourly_totals = (
    taxi.dropna(subset=[dropoff_col])
        .set_index(dropoff_col)
        .resample('h')[sum_cols]
        .sum()
        .sort_index()
 )


top_5_tip_hours = hourly_totals.nlargest(5, 'tip_amount')
top_5_tip_hours
```
